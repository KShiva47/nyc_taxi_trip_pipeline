```python
import json
import pandas as pd
import numpy as np
from pathlib import Path
import pyarrow.parquet as pq

BASE_DIR = Path.cwd().parent
print(BASE_DIR)
for path in Path(f"{BASE_DIR}/yellow_taxi_trip").iterdir():
    print(f"Analyzing metadata for file::: "+ path.name)
    parquet_file = pq.ParquetFile(path.absolute())
    # print("Schema structure::: "+ str(parquet_file.schema))
    # print(parquet_file.num_row_groups)
    schema = {field: str(type) for field, type in zip(parquet_file.schema_arrow.names, parquet_file.schema_arrow.types)}
    print("Schema :::", json.dumps(schema, indent = 4))
    print("\n")

    # print("Column length::: ", len(parquet_file.schema_arrow.names))
```

```python
# pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
for path in Path(f"{BASE_DIR}/yellow_taxi_trip").iterdir():
    print(f"Analyzing metadata for file::: "+ path.name)
    df = pd.read_parquet(path)
    print("Dataframe size:::", df.shape)
    print("Undocumented RatecodeId::: ", df[(df["RatecodeID"] == 99) | (df["RatecodeID"].isna())].shape)
    print(f"Negative charges::: ", df[df.select_dtypes(exclude= ["datetime64", "object"]).lt(0).any(axis = 1)].shape)
    print("Passenger count zero or NaN::: ", df[(df["passenger_count"] < 0.0 ) | (df["passenger_count"].isnull()) | (df["passenger_count"].isna()) | (df["passenger_count"] == 0.0)].shape)
    # print(df_fil.head())
    print("Trip distance zero::: ", df[(df["trip_distance"] == 0.0) | (df["trip_distance"].isna())].shape)
    print("Invalid datetime::: ", df[(df["tpep_pickup_datetime"].isna()) | (df["tpep_dropoff_datetime"].isna()) | (df["tpep_dropoff_datetime"] < df["tpep_pickup_datetime"])].shape)
    # print(df[df["PULocationID"] == 264].head())
    print("=================\n")
```
